```python
import matplotlib.pyplot as plt

hours=[1,3,5,7]
target=[12,23,32,34]
plt.scatter(hours,target,color="red",s=70,marker="o",label="Dot")
plt.title("Hours vs Target")
plt.xlabel("Hours")
plt.ylabel("Target")

plt.plot(hours,target,color="blue",label="Line plot")
plt.legend()
plt.grid(True)
plt.show()
```
